```python
# Brand-wise Sales Bar Chart
brand_sales = df['Brand'].value_counts()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=brand_sales.index,
    y=brand_sales.values
)

plt.title("Brand-wise Car Sales")
plt.xlabel("Car Brand")
plt.ylabel("Total Cars Sold")
plt.xticks(rotation=45)

plt.show()
```

```python
# Fuel Type Pie Chart
fuel_sales = df['Fuel_Type'].value_counts()
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    fuel_sales.values,
    labels=fuel_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Fuel Type Distribution")

plt.show()
```

```python
# Price Distribution Histogram

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Selling_Price',
    bins=20,
    kde=True
)

plt.title("Car Selling Price Distribution")
plt.xlabel("Selling Price")
plt.ylabel("Number of Cars")

plt.show()
```

```python
# Mileage Distribution Histogram 

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x='Mileage',
    bins=20,
    kde=True
)

plt.title("Car Mileage Distribution")
plt.xlabel("Mileage")
plt.ylabel("Number of Cars")

plt.show()
```

```python
# Transmission Count Plot

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Transmission'
)

plt.title("Transmission Type Distribution")
plt.xlabel("Transmission Type")
plt.ylabel("Number of Cars Sold")

plt.show()
```

```python
# Regional Sales Bar Chart

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Region',
    order=df['Region'].value_counts().index
)

plt.title("Region-wise Car Sales")
plt.xlabel("Region")
plt.ylabel("Number of Cars Sold")

plt.show()
```

```python
# Price vs Mileage Scatter Plot

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Mileage',
    y='Selling_Price'
)

plt.title("Price vs Mileage")
plt.xlabel("Mileage")
plt.ylabel("Selling Price")

plt.show()
```

```python
# Selling Price Box Plot

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y='Selling_Price'
)

plt.title("Selling Price Distribution")
plt.ylabel("Selling Price")

plt.show()
```

```python
# Correlation Heatmap

plt.figure(figsize=(8, 5))

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
